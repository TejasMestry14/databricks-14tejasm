# Load Dataset

```python
movies_df = spark.read.csv(
    "/Volumes/workspace/default/14tejasm_vol/movies.csv",
    header=True,
    inferSchema=True
)
display(movies_df)
```

Check No of Partitions Distribution

```python
# movies_df.rdd.getNumPartitions()

from pyspark.sql.functions import spark_partition_id

movies_with_pid = movies_df.withColumn(
    "partition_id",
    spark_partition_id()
)

display(movies_with_pid)
```

```python
movies_with_pid.select("partition_id").distinct().count()
```

Repartition

```python
repartitioned_df = movies_df.repartition(5)
```

```python
repartitioned_with_pid = repartitioned_df.withColumn(
    "partition_id",
    spark_partition_id()
)

display(repartitioned_with_pid)
```

```python
repartitioned_with_pid.select("partition_id").distinct().count()
```

```python
repartitioned_df.explain(True)
```

```python
coalesce_df = repartitioned_df.coalesce(2)
```

```python
coalesce_with_pid = coalesce_df.withColumn(
    "partition_id",
    spark_partition_id()
)

display(coalesce_with_pid)
```

```python
coalesce_df.explain()
```

## Real Life Scenario - Writing Files

```python
#writing without Coalesce

repartitioned_df.write.mode("overwrite").csv("/Volumes/workspace/default/14tejasm_vol/movies_output_repartitioned")
```

```python
#writing with Coalesce

coalesce_df.write.mode("overwrite").csv("/Volumes/workspace/default/14tejasm_vol/movies_output_coalesced")
```
