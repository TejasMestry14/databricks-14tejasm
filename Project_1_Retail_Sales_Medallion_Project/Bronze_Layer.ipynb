```python
%sql
CREATE TABLE workspace.default.walmart_dataset
USING DELTA
LOCATION 's3://databricks-demo-14tejasm/bronze/'
AS
SELECT
    *,
    current_timestamp() as created_at,
    input_file_name() as source_path,
    uuid() as record_id
FROM read_files(
    's3://databricks-demo-14tejasm/',
    format => 'csv',
    header => true,
    inferSchema => true
);
```
